```python
import pandas as pd
import os
```

**Load Datasets**

```python
parquet_directory = 'data'

parquet_files = [f for f in os.listdir(parquet_directory) if f.endswith('.parquet')]

dfs = {}

for file in parquet_files:
    file_path = os.path.join(parquet_directory, file)
    dfs[file] = pd.read_parquet(file_path)
```

```python
dfs.keys()
```

**Explore Metadata Dataset**

**Giải thích:** `.dtypes` cho biết kiểu từng cột: biến khí tượng chủ yếu `float32`, `timestamp` là datetime, `location_id` là chuỗi để liên kết với metadata.

**Giải thích:** Lấy `timestamp` nhỏ nhất và lớn nhất trong `weather.parquet` để biết **khoảng thời gian** dữ liệu thời tiết (có thể dài hơn hoặc khác demand — cần căn chỉnh khi ghép nối chuỗi).
